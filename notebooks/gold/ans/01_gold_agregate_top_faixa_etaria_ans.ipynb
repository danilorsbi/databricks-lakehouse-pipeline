```python
%sql
-- utilizando o catalogo e schema padrao

USE CATALOG bmg_saude_desafio;
USE SCHEMA gold;
```

```python
%sql
-- criando tabela beneficiarios por faixa etaria

CREATE OR REPLACE TABLE gold.beneficiarios_por_faixa_etaria
USING DELTA
COMMENT 'Total de beneficiarios ativos por faixa etaria'
AS
SELECT
  de_faixa_etaria,
  SUM(qt_beneficiario_ativo) AS total_beneficiarios_ativos
FROM silver.beneficiarios_ans
GROUP BY de_faixa_etaria
ORDER BY total_beneficiarios_ativos DESC;
```

```python
%sql
-- select por faixa hetária

SELECT de_faixa_etaria, total_beneficiarios_ativos
FROM gold.beneficiarios_por_faixa_etaria
ORDER BY total_beneficiarios_ativos DESC
LIMIT 1;
 
-- COMMAND ----------
 
-- MAGIC %md
-- MAGIC Resultado esperado: faixa **"35 a 39 anos"**, com **27.446**
-- MAGIC beneficiários ativos.
```
