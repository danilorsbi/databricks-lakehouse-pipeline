```python
%sql
-- utilizando o catalogo e schema padrao

USE CATALOG bmg_saude_desafio;
USE SCHEMA gold;
```

```python
%sql
-- criando tabela beneficiarios_por_municipio

CREATE OR REPLACE TABLE gold.beneficiarios_por_municipio
USING DELTA
COMMENT 'Total de beneficiarios ativos por municipio, ordem decrescente'
AS
SELECT
  nm_municipio,
  SUM(qt_beneficiario_ativo) AS total_beneficiarios_ativos
FROM silver.beneficiarios_ans
GROUP BY nm_municipio
ORDER BY total_beneficiarios_ativos DESC;
```

```python
%sql

SELECT nm_municipio, total_beneficiarios_ativos
FROM gold.beneficiarios_por_municipio
ORDER BY total_beneficiarios_ativos DESC;

```
